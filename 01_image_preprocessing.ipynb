```python
# ── 3. Configuration ──────────────────────────────────────────────────────────
import os

# Adjust these paths to match your Drive layout
INPUT_DIR  = '/content/drive/MyDrive/vp_lab/raw'       if IN_COLAB else './data/raw'
OUTPUT_DIR = '/content/drive/MyDrive/vp_lab/processed' if IN_COLAB else './data/processed'

TARGET_SIZE    = (512, 512)  # (width, height)
APPLY_FACES    = True        # blur detected faces
APPLY_TATTOOS  = True        # mask detected tattoo regions
BLUR_STRENGTH  = 51          # Gaussian kernel size (odd number)

print(f'Input  : {INPUT_DIR}')
print(f'Output : {OUTPUT_DIR}')
```

```python
# ── 5. Run preprocessing ──────────────────────────────────────────────────────
from preprocessing import adaptar_imagenes_crudas

exported = adaptar_imagenes_crudas(
    input_dir=INPUT_DIR,
    output_dir=OUTPUT_DIR,
    target_size=TARGET_SIZE,
    apply_face_mask=APPLY_FACES,
    apply_tattoo_mask=APPLY_TATTOOS,
    blur_strength=BLUR_STRENGTH,
)
print(f'\n✅  {len(exported)} images exported to {OUTPUT_DIR}')
```

```python
# ── 6. Quick visual preview ───────────────────────────────────────────────────
import random
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

sample = random.sample(exported, min(6, len(exported)))
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
for ax, p in zip(axes.flat, sample):
    ax.imshow(Image.open(p))
    ax.set_title(Path(p).name, fontsize=8)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
